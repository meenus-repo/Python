```python
### The `sys` Module
#The `sys` module provides access to variables and functions related to the Python interpreter.

#**Common uses:**
#- Command-line arguments (`sys.argv`)
#- Exiting programs (`sys.exit`)
#- Python version info (`sys.version`)
```

```python
import sys

print("Python Version:", sys.version)
print("Executable Path:", sys.executable)

# Command line arguments demo (first is script name)
print("Command-line args:", sys.argv)
```

```python
### The `os` Module
#The `os` module allows interaction with the operating system.

#Common uses:
#- Getting current directory
#- Listing files
#- Creating/removing directories
#- Accessing environment variables
```

```python
import os

print("Current Working Directory:", os.getcwd())
print("Files in current directory:", os.listdir())

# Uncomment to create/remove a directory
# os.mkdir("demo_dir")
# os.rmdir("demo_dir")
```

```python
### Handling Errors with try-except
#Errors (exceptions) can occur during runtime. To prevent the program from crashing, we use `try-except` blocks.
```

```python
try:
    num = int("abc")  # This will raise ValueError
except ValueError as e:
    print("Handled Error:", e)

print("Program continues...")
```

```python
# We can also handle **multiple exceptions**:
```

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("You can't divide by zero!")
except Exception as e:
    print("Other error:", e)
```

```python
### Debugging Basics
#- **print() debugging**: Add print statements to trace variable values.
#- **Using `pdb` module**: Python Debugger lets you step through code.
#- **IDE Debuggers**: Most IDEs (PyCharm, VSCode, Jupyter) have built-in debugging tools.
```

```python
import pdb

def divide(a, b):
    pdb.set_trace()  # Debug breakpoint
    return a / b

# Uncomment to test debugging
# divide(10, 0)
```
