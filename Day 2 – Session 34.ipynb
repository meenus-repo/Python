````python
### The `datetime` Module
#The `datetime` module is used to work with dates and times.
````

````python
import datetime

now = datetime.datetime.now()
print("Current Date & Time:", now)
print("Today's Date:", now.date())
print("Current Time:", now.time())

# Creating specific date
d = datetime.date(2025, 1, 1)
print("Custom Date:", d)
````

````python
### Project Structure Example
#```
#myproject/
#│── main.py
#│── requirements.txt
#└── mypackage/
#    ├── __init__.py
#    ├── module1.py
#    ├── module2.py
#```
#- `main.py`: entry point of program
#- `requirements.txt`: lists dependencies
#- `mypackage`: contains related modules
````

````python
### Packages Recap
#- A **module** = single `.py` file
#- A **package** = directory containing `__init__.py`
#- Packages allow hierarchical structure (sub-packages)

#Example:
#```
#mypackage/
#    __init__.py
#    module1.py
#    subpackage/
#        __init__.py
#        module2.py
#```
````

````python
### Interview Notes on `__init__.py`
#- `__init__.py` itself is a **module** (like any `.py` file)
#- It makes the folder a **package**
#- Can be empty, or run initialization code
#- Can expose selected functions/classes at the package level
#- Modern Python (3.3+) supports **namespace packages** without `__init__.py`, but it’s still widely used
````
