```python
df1 = pd.DataFrame({'ID': [1, 2], 'Name': ['Alice', 'Bob']})
df2 = pd.DataFrame({'ID': [1, 2], 'Age': [25, 30]})

merged = pd.merge(df1, df2, on='ID')
print("Merged:")
print(merged)

concatenated = pd.concat([df1, df2], axis=1)
print("Concatenated:")
print(concatenated)
```
