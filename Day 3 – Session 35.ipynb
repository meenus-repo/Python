```python
### Creating NumPy Arrays
```

```python
import numpy as np

# From Python list
arr = np.array([1, 2, 3, 4])
print("Array:", arr)

# Zeros and Ones
print("Zeros:", np.zeros((2, 3)))
print("Ones:", np.ones((2, 3)))

# Range
print("Arange:", np.arange(0, 10, 2))

# Random
print("Random:", np.random.rand(2, 2))
```

```python
### Vectorized Operations
#NumPy allows performing operations on arrays **element-wise** without explicit loops.
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Addition:", a + b)
print("Multiplication:", a * b)
print("Square:", a ** 2)
```

```python
### Broadcasting
#Broadcasting allows NumPy to perform arithmetic between arrays of different shapes.
```

```python
a = np.array([1, 2, 3])
print(a + 10)

matrix = np.ones((3,3))
print(matrix + a)
```
