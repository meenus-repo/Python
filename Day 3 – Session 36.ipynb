```python
### Reading Data into DataFrames
```

```python
import pandas as pd

# Creating DataFrame from dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['NY', 'LA', 'Chicago']
}
df = pd.DataFrame(data)
print(df)
```

```python
### Exploring Data
#Useful methods to understand data:
#- `head()`
#- `info()`
#- `describe()`
```

```python
print(df.head())
print(df.info())
print(df.describe())
```

```python
### Series vs DataFrame
#- **Series**: 1D labeled array (like a single column)
#- **DataFrame**: 2D labeled data structure (like a table)
```

```python
series = df['Age']
print("Series:")
print(series)

print("DataFrame column type:", type(series))
print("DataFrame type:", type(df))
```
