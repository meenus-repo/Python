````python
### Wide vs Long Format
- **Wide format**: each subject has a single row with multiple columns of data.
- **Long format**: each observation is a separate row.

**Example:**

Wide:
```
Name   Age   City
Alice   25   NY
Bob     30   LA
```

Long:
```
Name   Variable   Value
Alice  Age        25
Alice  City       NY
Bob    Age        30
Bob    City       LA
```
````
