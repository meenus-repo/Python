```python
### Building and Editing Arrays
```

```python
arr = np.array([1, 2, 3, 4, 5, 6])
print("Original:", arr)

# Editing
arr[0] = 10
print("Edited:", arr)
```

```python
### Reshaping and Resizing
```

```python
matrix = arr.reshape(2, 3)
print("Reshaped:")
print(matrix)

arr.resize((2, 2))
print("Resized:")
print(arr)
```

```python
### Matrix Operations
```

```python
m1 = np.array([[1, 2], [3, 4]])
m2 = np.array([[5, 6], [7, 8]])

print("Matrix Addition:")
print(m1 + m2)

print("Matrix Multiplication:")
print(np.dot(m1, m2))
```
