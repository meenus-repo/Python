```python
### What is a Module?
#  A **module** in Python is simply a file that contains Python code. It may include functions, classes, or variables. Modules allow us to organize and reuse code across multiple programs.

#  **Example:** `math.py` is a module that provides mathematical functions.
```

```python
import math

# Using math module functions
print("Square root of 16:", math.sqrt(16))
print("Value of pi:", math.pi)
print("Cosine of 0:", math.cos(0))
```

```python
### Ways to Import Modules
# 1. `import module_name`
# 2. `import module_name as alias`
# 3. `from module_name import function`
# 4. `from module_name import *`

⚠️ **Note:** Avoid using `*` (wildcard import) in large projects.
```

```python
import math as m
print(m.factorial(5))

from math import pow
print(pow(2, 3))
```

```python
### The `random` Module
#The `random` module is part of Python’s standard library and is used for generating random numbers.
```

```python
import random

print("Random float between 0 and 1:", random.random())
print("Random integer between 1 and 10:", random.randint(1, 10))
print("Random choice from a list:", random.choice(['apple', 'banana', 'cherry']))
print("Shuffling a list:")
items = [1, 2, 3, 4, 5]
random.shuffle(items)
print(items)
```

```python
### Difference Between Modules and Packages
#- **Module:** A single `.py` file containing code (functions, classes, variables).
#- **Package:** A collection of modules in a directory with an `__init__.py` file (makes it recognizable as a package).

#**Example:**
#- Module: `math.py`
#- Package: `numpy` (contains many modules inside)
```
